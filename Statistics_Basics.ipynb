````python


**Question 1: What is the difference between descriptive statistics and inferential statistics? Explain with examples.**


**Answer:**

* **Descriptive Statistics:** Focuses on organizing, summarizing, and presenting data in a meaningful way. It describes the data that is already known and does not make predictions beyond that dataset.
* *Example:* Calculating the average test score of students in a specific classroom, or creating a bar chart showing the number of cars sold by a dealership in a month.


* **Inferential Statistics:** Uses a sample of data to draw conclusions, make predictions, or infer properties about a larger population.
* *Example:* Surveying 1,000 voters to predict the outcome of a national election, or testing a new drug on a sample group to infer its effectiveness on the general population.



---

**Question 2: What is sampling in statistics? Explain the differences between random and stratified sampling.**


**Answer:**

* **Sampling:** The process of selecting a subset of individuals or observations from a larger population to estimate characteristics of the whole population.
* **Random Sampling:** Every member of the population has an equal and independent chance of being selected. It is simple but might not perfectly represent underlying subgroups.
* **Stratified Sampling:** The population is first divided into distinct, non-overlapping subgroups (called strata) based on specific characteristics (e.g., age, gender, income). Random samples are then drawn from each stratum proportionally. This ensures that minority subgroups are adequately represented.

---

**Question 3: Define mean, median, and mode. Explain why these measures of central tendency are important.**


**Answer:**

* **Mean:** The mathematical average of a dataset, calculated by adding all values and dividing by the total number of values.
* **Median:** The middle value of a dataset when it is ordered from least to greatest. If there is an even number of observations, it is the average of the two middle values.
* **Mode:** The value that appears most frequently in a dataset.
* **Importance:** These measures summarize a large set of data into a single, representative value. They help identify the "center" or typical behavior of a dataset, which is crucial for comparison and foundational for more complex statistical analysis.

---

**Question 4: Explain skewness and kurtosis. What does a positive skew imply about the data?**


**Answer:**

* **Skewness:** Measures the asymmetry of a distribution around its mean. A perfectly symmetrical distribution (like a normal curve) has a skewness of zero.
* **Kurtosis:** Measures the "tailedness" of a distribution. High kurtosis indicates heavy tails (more outliers), while low kurtosis indicates light tails (fewer outliers) compared to a normal distribution.
* **Positive Skew Implication:** A positive (or right) skew implies that the tail on the right side of the distribution is longer or fatter than the left side. It indicates that the bulk of the data is concentrated on the lower end, with a few exceptionally high values dragging the mean upward. In a positive skew, the Mean > Median > Mode.

---

### **Code Cell 1: Mean, Median, and Mode (Q5)**

*(Copy this code into a Colab Code cell)*

```python
# Question 5: Compute mean, median, and mode[cite: 6]
import statistics

numbers = [12, 15, 12, 18, 19, 12, 20, 22, 19, 19, 24, 24, 24, 26, 28]

# Computations
mean_val = statistics.mean(numbers)
median_val = statistics.median(numbers)
mode_val = statistics.multimode(numbers) # multimode handles multiple modes

print("--- Question 5 Output ---")
print(f"Dataset: {numbers}")
print(f"Mean: {mean_val:.2f}")
print(f"Median: {median_val}")
print(f"Mode(s): {mode_val}")

```

---

### **Code Cell 2: Covariance and Correlation (Q6)**

*(Copy this code into a Colab Code cell)*

```python
# Question 6: Compute covariance and correlation coefficient[cite: 6]
import numpy as np

list_x = [10, 20, 30, 40, 50]
list_y = [15, 25, 35, 45, 60]

# Computations
# np.cov returns a covariance matrix, [0][1] gets the covariance between x and y
covariance = np.cov(list_x, list_y)[0][1]

# np.corrcoef returns a correlation matrix, [0][1] gets the correlation between x and y
correlation = np.corrcoef(list_x, list_y)[0][1]

print("--- Question 6 Output ---")
print(f"Covariance: {covariance:.2f}")
print(f"Correlation Coefficient: {correlation:.4f}")

```

---

### **Code Cell 3: Boxplot and Outliers (Q7)**

*(Copy this code into a Colab Code cell)*

```python
# Question 7: Draw a boxplot and identify outliers[cite: 6]
import matplotlib.pyplot as plt
import numpy as np

data = [12, 14, 14, 15, 18, 19, 19, 21, 22, 22, 23, 23, 24, 26, 29, 35]

# Calculate IQR for Outlier Identification
q1 = np.percentile(data, 25)
q3 = np.percentile(data, 75)
iqr = q3 - q1

lower_bound = q1 - (1.5 * iqr)
upper_bound = q3 + (1.5 * iqr)

outliers = [x for x in data if x < lower_bound or x > upper_bound]

print("--- Question 7 Output ---")
print(f"Q1: {q1}, Q3: {q3}, IQR: {iqr}")
print(f"Lower Bound: {lower_bound}, Upper Bound: {upper_bound}")
print(f"Identified Outliers: {outliers}")
print("\nExplanation: The Interquartile Range (IQR) method defines outliers as points that fall below Q1 - 1.5*IQR or above Q3 + 1.5*IQR. In this dataset, 35 exceeds the upper bound, making it an outlier.")

# Draw Boxplot
plt.figure(figsize=(8, 4))
plt.boxplot(data, vert=False, patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Boxplot of Numeric List")
plt.xlabel("Values")
plt.show()

```

---

### **Text Cell 2: E-commerce and Survey Analysis (Q8 - Q9)**

*(Copy this text into a Colab Markdown cell)*

**Question 8: E-commerce Advertising Spend vs. Daily Sales**

* **Explanation:** Covariance determines the *direction* of the relationship between advertising spend and sales. A positive covariance indicates that as ad spend increases, sales tend to increase. The correlation coefficient goes a step further by measuring the *strength* and standardized direction of this relationship on a scale of -1 to 1. A correlation close to 1 would give the marketing team high confidence that increased ad spend strongly predicts higher daily sales.

**Question 9: Customer Satisfaction Survey Distribution**

* **Explanation:**
* *Mean:* Used to find the average satisfaction score, giving a quick pulse on overall customer sentiment.
* *Median:* Used alongside the mean to check for skewness. If the mean is significantly lower than the median, low-score outliers might be pulling the average down.
* *Standard Deviation:* Used to understand the spread or consistency of the scores. A low standard deviation means most customers agree, while a high one indicates polarizing opinions.
* *Histogram:* The best visualization to see the frequency of each score bracket, revealing if the data is normally distributed, skewed, or bimodal (two distinct groups of opinions).



---

### **Code Cell 4: Python scripts for Q8 and Q9**

*(Copy this code into a Colab Code cell)*

```python
# Question 8 & 9 Implementations[cite: 6]
import numpy as np
import matplotlib.pyplot as plt

print("--- Question 8 Output: E-commerce Correlation ---")
advertising_spend = [200, 250, 300, 400, 500]
daily_sales = [2200, 2450, 2750, 3200, 4000]

correlation_sales = np.corrcoef(advertising_spend, daily_sales)[0][1]
print(f"Correlation between Ad Spend and Daily Sales: {correlation_sales:.4f}")
print("Result: A near-perfect positive correlation (close to 1.0) means higher ad spend strongly drives higher sales.")

print("\n--- Question 9 Output: Survey Histogram ---")
survey_scores = [7, 8, 5, 9, 6, 7, 8, 9, 10, 4, 7, 6, 9, 8, 7]

mean_score = np.mean(survey_scores)
std_dev = np.std(survey_scores)

print(f"Survey Mean: {mean_score:.2f}")
print(f"Survey Standard Deviation: {std_dev:.2f}")

# Histogram visualization
plt.figure(figsize=(8, 5))
plt.hist(survey_scores, bins=np.arange(3.5, 11.5, 1), edgecolor='black', color='coral', alpha=0.8)
plt.title("Customer Satisfaction Survey Distribution")
plt.xlabel("Satisfaction Score (1-10)")
plt.ylabel("Frequency")
plt.xticks(range(4, 11))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

```
````
